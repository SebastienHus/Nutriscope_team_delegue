```python
#%pip install duckdb pandas pyarrow

from pathlib import Path

file = Path("../data/en.openfoodfacts.org.products.csv")

print(file.exists())


import duckdb

con = duckdb.connect()

con.execute("""
CREATE VIEW food AS
SELECT *
FROM read_csv_auto('../data/en.openfoodfacts.org.products.csv');
""")
```

```python
df = con.sql("""
    SELECT main_category, product_name, categories, categories_tags, completeness, no_nutrition_data
    FROM food
             LIMIT 50
""").df()

df.head()
```

```python
schema_df = con.sql("""
DESCRIBE
SELECT *
FROM food
""").df()

schema_df.to_csv("../data/tp2/schema_openfoodfacts.csv", index=False)
```
